```polyglot-notebook
using System;
using System.Collections.Generic;
using System.Linq;

// ====== Интерфейс ======
public interface IInstallable
{
    void Install();
}

// ====== Абстрактный базовый класс ======
public abstract class Component : IInstallable
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    protected Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    // Виртуальный метод — будет переопределён в наследниках
    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, Цена: {Price}₽");
    }

    // Перегрузка ToString()
    public override string ToString()
    {
        return $"{Name} ({Price}₽)";
    }

    // Абстрактная реализация интерфейса
    public abstract void Install();

    // Перегрузка операторов сравнения
    public static bool operator >(Component c1, Component c2) => c1.Price > c2.Price;
    public static bool operator <(Component c1, Component c2) => c1.Price < c2.Price;
}

// ====== Конкретные классы ======
public class CPU : Component
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores) : base(name, price)
    {
        Cores = cores;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"CPU: {Name}, Ядер: {Cores}, Цена: {Price}₽");
    }

    public override void Install()
    {
        Console.WriteLine($"Установка CPU {Name}...");
    }
}

public class RAM : Component
{
    public int Capacity { get; set; }

    public RAM(string name, decimal price, int capacity) : base(name, price)
    {
        Capacity = capacity;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"RAM: {Name}, Объем: {Capacity} ГБ, Цена: {Price}₽");
    }

    public override void Install()
    {
        Console.WriteLine($"Установка RAM {Name}...");
    }
}

public class GPU : Component
{
    public int Memory { get; set; }

    public GPU(string name, decimal price, int memory) : base(name, price)
    {
        Memory = memory;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"GPU: {Name}, Память: {Memory} ГБ, Цена: {Price}₽");
    }

    public override void Install()
    {
        Console.WriteLine($"Установка GPU {Name}...");
    }
}

public class HDD : Component
{
    public int Capacity { get; set; }

    public HDD(string name, decimal price, int capacity) : base(name, price)
    {
        Capacity = capacity;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"HDD: {Name}, Объем: {Capacity} ГБ, Цена: {Price}₽");
    }

    public override void Install()
    {
        Console.WriteLine($"Установка HDD {Name}...");
    }
}

// ====== Класс Computer ======
public class Computer
{
    private List<Component> components = new List<Component>();

    // Добавление компонента (Generic)
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        component.Install();
        components.Add(component);
        Console.WriteLine($"Компонент {component.Name} добавлен в компьютер.");
    }

    // Удаление компонента (Generic)
    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
            Console.WriteLine($"Компонент {component.Name} удален из компьютера.");
        else
            Console.WriteLine($"Компонент {component.Name} не найден.");
    }

    // Подсчет общей цены
    public decimal GetTotalPrice() => components.Sum(c => c.Price);

    // Вывод всех компонентов
    public void ShowComponents()
    {
        Console.WriteLine("\nТекущая конфигурация:");
        foreach (var c in components)
            c.GetInfo();
        Console.WriteLine($"Общая стоимость: {GetTotalPrice()}₽");
    }
}
// ====== Фабрика ======
public static class ComputerFactory
{
    public static Computer CreateBasicPC()
    {
        var pc = new Computer();
        pc.AddComponent(new CPU("Intel i5", 15000, 6));
        pc.AddComponent(new RAM("Kingston 16GB", 5000, 16));
        pc.AddComponent(new HDD("Seagate 1TB", 4000, 1000));
        return pc;
    }

    public static Computer CreateGamingPC()
    {
        var pc = new Computer();
        pc.AddComponent(new CPU("AMD Ryzen 7", 25000, 8));
        pc.AddComponent(new RAM("Corsair 32GB", 10000, 32));
        pc.AddComponent(new GPU("NVIDIA RTX 4070", 60000, 12));
        pc.AddComponent(new HDD("Samsung SSD 2TB", 10000, 2000));
        return pc;
    }
}

// ====== Главная программа ======


        Console.OutputEncoding = System.Text.Encoding.UTF8;
        Console.WriteLine("=== Демонстрация полиморфизма и Generic ===");

        // Создание базового компьютера
        var basic = ComputerFactory.CreateBasicPC();
        basic.ShowComponents();

        Console.WriteLine("\n=== Создание игрового компьютера ===");
        var gaming = ComputerFactory.CreateGamingPC();
        gaming.ShowComponents();

        Console.WriteLine("\n=== Сравнение компонентов ===");
        var cpu1 = new CPU("Intel i7", 50000, 6);
        var cpu2 = new CPU("AMD Ryzen 7", 25000, 8);
        Console.WriteLine(cpu1 > cpu2
            ? $"{cpu1.Name} дороже {cpu2.Name}"
            : $"{cpu1.Name} дешевле {cpu2.Name}");

        Console.WriteLine("\n=== Конец демонстрации ===");
```
